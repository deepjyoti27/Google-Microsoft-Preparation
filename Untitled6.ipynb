```python
#small ML style example
import numpy as np
marks=np.array([60,80,90,100,85])
average=np.mean(marks)
adjusted_marks=marks+5
print("Average:",average)
print("Adjusted Marks:",adjusted_marks)
```

```python
import numpy as np

arr = np.array([[1, 2, 3],
                [4, 5, 6]])

print(arr.ndim)   # 2
print(arr.shape)  # (2, 3)
print(arr.size)   # 6
```
